```python
class Pokemon:
    def __init__(self, name, type_class, hp=100, pp=20):
        self.name = name
        self.type_class = type_class
        self.hp = hp
        self.pp = pp

pikachu = Pokemon(name="Pikachu", type_class="Electric")
squirtle = Pokemon(name="Squirtle", type_class="Water")

def print_attack(move_func):
    def move_func_with_print(*args):
        print(f"{move_func.__name__} was used!")
        return move_func(*args)
    return move_func_with_print

move_log = []

def log_move(move_func):
    def move_func_with_log(*args):
        move_log.append(move_func.__name__)
        return move_func(*args)
    return move_func_with_log

def pp_decorator_factory(pp_cost):
    def pp_decorator(move_func):
        def move_func_with_pp(*args):
            print(f"Using it cost {pp_cost}!")
            return move_func(*args)
        return move_func_with_pp
    return pp_decorator

@print_attack
@log_move
@pp_decorator_factory(pp_cost=2)
def use_thunderbolt(user, target):
    target.hp -= 10

use_thunderbolt(pikachu, squirtle)

print(pikachu.hp)
print(squirtle.hp)
print(move_log)
```

```python
class MoveClass:
    def __init__(self, move_func, name, type):
        self.name = name
        self.type = type
        self.move_func = move_func

    def use(self, user, target):
        self.move_func(user, target)
        print(f"It's a {self.type} move!")
    
def expand_move_func_to_class(name, type):
    def move_func_class(move_func):
        return MoveClass(move_func, name, type)
    return move_func_class

@expand_move_func_to_class(name="surf", type="water")
@print_attack
def use_surf_logic(user, target):
    target.hp -= 10

use_surf_logic.use(squirtle, pikachu)
print(pikachu.hp)
print(squirtle.hp)
```
